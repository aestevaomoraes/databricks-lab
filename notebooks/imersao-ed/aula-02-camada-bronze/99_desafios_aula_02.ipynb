# ⚙️ Desafio 1 - Validando dados da Camada Bronze

## Contexto

Na Aula 2 realizamos a ingestão dos dados para a camada Bronze da Arquitetura Medalhão.

A camada Bronze tem como objetivo armazenar os dados o mais próximo possível da origem, preservando o conteúdo original para futuras transformações nas camadas Silver e Gold.

Antes de avançar para as próximas etapas do pipeline, é importante validar se a ingestão foi realizada corretamente.

## Objetivo

Verificar:

- Quantidade de registros armazenados.
- Estrutura da tabela (colunas disponíveis).
- Primeiros registros carregados.

Essa validação permite garantir que os dados foram ingeridos corretamente sem realizar qualquer alteração na tabela.

## Tabelas analisadas

- voo_bem.bronze.aerodromos
- voo_bem.bronze.empresas_nacionais
- voo_bem.bronze.empresas_estrangeiras

```python
tabelas = [
    "voo_bem.bronze.aerodromos",
    "voo_bem.bronze.empresas_nacionais",
    "voo_bem.bronze.empresas_estrangeiras"
]

for tabela in tabelas:

    df = spark.table(tabela)

    print("=" * 80)
    print(f"Tabela: {tabela}")

    print(f"\nQuantidade de registros: {df.count():,}")

    print("\nColunas disponíveis:")

    for coluna in df.columns:
        print(f"  • {coluna}")

    print("\nPrimeiros registros:")

    display(df.limit(5))
```

## Resultado

A validação confirmou que as tabelas foram carregadas corretamente na camada Bronze.

### Quantidade de registros

| Tabela | Registros |
|---------|---------:|
| aerodromos | 496 |
| empresas_nacionais | 729 |
| empresas_estrangeiras | 148 |

### Estrutura

As colunas foram carregadas conforme esperado para cada tabela, incluindo as colunas de auditoria:

- _arquivo_origem
- _ingerido_em

### Amostra dos dados

Os primeiros registros retornados apresentaram conteúdo consistente com a origem dos dados e sem evidência de falhas na ingestão.

## Estrutura da tabela

Para validar a ingestão dos dados na camada Bronze, vamos inspecionar as colunas disponíveis em cada tabela.

Essa verificação permite confirmar se os campos esperados foram carregados corretamente e se a estrutura está pronta para as próximas etapas do pipeline.

```python
tabelas = [
    "voo_bem.bronze.aerodromos",
    "voo_bem.bronze.empresas_nacionais",
    "voo_bem.bronze.empresas_estrangeiras"
]

for tabela in tabelas:
    print("=" * 80)
    print(f"Tabela: {tabela}")
    print("\nColunas disponíveis:")

    for coluna in spark.table(tabela).columns:
        print(f"• {coluna}")

    print()
```

## Conclusão

A inspeção das colunas confirmou que as tabelas da camada Bronze foram criadas corretamente.

Além dos campos de negócio, também foram carregadas as colunas de auditoria (_arquivo_origem e _ingerido_em), importantes para rastreabilidade e governança dos dados ao longo do pipeline.

## Primeiros registros armazenados

Após validar a quantidade de registros e a estrutura das tabelas, vamos inspecionar uma amostra dos dados carregados.

O objetivo é confirmar visualmente que os registros foram ingeridos corretamente e que os valores apresentados são consistentes com os dados de origem.

Essa é uma prática comum em Engenharia de Dados para validar a qualidade da ingestão antes de avançar para as próximas camadas do pipeline.

```python
tabelas = [
    "voo_bem.bronze.aerodromos",
    "voo_bem.bronze.empresas_nacionais",
    "voo_bem.bronze.empresas_estrangeiras"
]

for tabela in tabelas:

    print("=" * 80)
    print(f"Primeiros registros da tabela: {tabela}")

    display(
        spark.table(tabela)
        .limit(5)
    )
```
